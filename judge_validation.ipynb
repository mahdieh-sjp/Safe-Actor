```python
import glob
import pandas as pd
from utils.seeds import initialize_seeds
from itertools import cycle
import random
import json
```

```python
DATA_DIR = "./golden-dataset/data/"
files = glob.glob(f"{DATA_DIR}*_dataset.csv")

df_list = [pd.read_csv(f) for f in files if "empty" not in f]
merged_df = pd.concat(df_list, axis=0, ignore_index=True)
test_df = merged_df[merged_df["split"] == "test"]
```

```python
test_df
```

```python
personas = test_df.persona.unique().tolist(); personas
```

```python
models = glob.glob("generations/*")
```

```python
model
```

```python
samples = []
```

```python
persona_cycle = cycle(personas)
model_cycle = cycle(models)
```

```python
initialize_seeds()
```

```python
for _ in range(100):
    model = next(model_cycle)
    persona = next(persona_cycle)
    prompts = test_df[test_df['persona'] == persona]["prompt"].tolist()
    index = random.choice(range(len(prompts)))
    prompt = prompts[index]
    response = json.load(open(f"{model}/{persona.replace(" ", "_")}.json", "r"))[index]
    samples.append({"model": model, "persona": persona, "prompt": prompt, "response": response})
```

```python
df = pd.DataFrame(samples)
```
